TASK 1 

```python
import random
import numpy as np

np.random.seed(0)

readings = np.random.normal(loc=50, scale=15, size=(360,4))

print("Shape:", readings.shape)
print("Data type:", readings.dtype)
print("First three rows:\n", readings[:3])
```

TASK 2

```python
min__vals = readings.min(axis=0)
max__vals = readings.max(axis=0)

scaled_readings = (readings - min__vals) / (max__vals - min__vals)
```

```python
col_means = readings.mean(axis=0)
centered_readings = readings - col_means
```

```python
print("Original shape:", readings.shape)
print("Scaled shape:", scaled_readings.shape)
print("Centered shape:", centered_readings.shape)
```

TASK 3

```python
outlier_mask = (readings > 80) | (readings < 20)
outlier_count = np.sum(outlier_mask)
total_readings = readings.size
outlier_percentage = ( outlier_count / total_readings ) * 100

print("Number of outliers:", outlier_count)
print("Percentage of outliers: {:.2f}%".format(outlier_percentage))

cleaned_readings = readings.copy()
cleaned_readings[outlier_mask] = np.nan


nan_count = np.isnan(cleaned_readings).sum()
print("Number of np.nan in cleaned array:", nan_count)
```

TASK 4

```python
sensor_means = readings.mean(axis=0)
sensor_stds = readings.std(axis=0)
time_means = readings.mean(axis=1)
```

TASK 5

```python
report_dict = {
    "total_readings": total_readings,
    "outlier_count": outlier_count,
    "outlier_percentage": outlier_percentage,
    "sensor_means": sensor_means,
    "sensor_stds": sensor_stds, 
}

print(f"Total readings count is: {total_readings} \nTotal outlier count is: {outlier_count} \nOutlier percentage is: {outlier_percentage:0.2f} \nSensor means are: {sensor_means} \nSensor stds are: {sensor_stds}")
print(f"Total readings count is the same as readings size: {total_readings == readings.size}")
```

TASK 6

```python
np.random.seed(0)

sim = np.random.normal(loc=0, scale=1, size=(1000, 6))

print(f"Sim shape is: {sim.shape}")
print(f"Sim type is: {sim.dtype}")
print(f"First three rows are: \n{sim[:3]}")
```

TASK 7

```python
scenario_shift = np.array([-0.3, -0.1, 0.0, 0.1, 0.2, 0.4])

adjusted_sim = sim + scenario_shift


print(f"Shapes are the same: {sim.shape == adjusted_sim.shape}")

original_means = sim.mean(axis=0)
adjusted_means = adjusted_sim.mean(axis=0)

print(f"Mean shift is: {original_means-adjusted_means}")
```

TASK 8

```python

scenario_means = adjusted_sim.mean(axis=0)
scenario_ranking = np.argsort(scenario_means)

partitioned_means = np.partition(scenario_means, -2)
top_two_partitioned = np.sort(partitioned_means[-2:]) 

full_sorted_means = np.sort(scenario_means)
top_two_sorted = full_sorted_means[-2:]

print(f"Scenario Means: {scenario_means}")
print(f"Ranking (Indices): {scenario_ranking}")
print(f"Top 2 match: {np.allclose(top_two_partitioned, top_two_sorted)}")
```

TASK 9

```python

np.random.seed(0)
sim_2 = np.random.normal(loc=0, scale=1, size=(1000, 6))

np.random.seed(42)
sim_3 = np.random.normal(loc=0, scale=1, size=(1000, 6))

match_1_2 = np.array_equal(sim, sim_2)
match_1_3 = np.array_equal(sim, sim_3)

print(f"Does sim_2 match sim? {match_1_2}")
print(f"Does sim_3 match sim? {match_1_3}")
print(f"Number of identical elements in sim vs sim_3: {np.sum(sim == sim_3)} out of {sim.size}")
```

TASK 10

```python

top_two_indices = scenario_ranking[-2:]
top_two_values = scenario_means[top_two_indices]

validation_report = {
    "shape": adjusted_sim.shape,
    "scenario_means": scenario_means,
    "top_two_indices": top_two_indices,
    "top_two_values": top_two_values
}

val_len_means = len(scenario_means) == 6
val_len_indices = len(top_two_indices) == 2

report_str = f"""
==========================================================
                 VALIDATION REPORT PART 2                 
==========================================================                 
Array Shape:         {validation_report['shape']}
Scenario Means:      {validation_report['scenario_means']}
Top 2 Indices:       {validation_report['top_two_indices']}
Top 2 Values:        {validation_report['top_two_values']}

VALIDATION CHECKS:
- Scenario means count is 6: {val_len_means}
- Top two indices count is 2: {val_len_indices}
==========================================================
"""
print(report_str)
```
